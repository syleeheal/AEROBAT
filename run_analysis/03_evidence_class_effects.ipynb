# Behavior-evaluation-rubric evidence-class effects
Rebuilds the per-evidence-class effect matrices used to qualify aggregate behavioral findings in Appendix D.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'run_analysis': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from aerobat.analysis.paper import evidence_class_effects, plot_evidence_heatmaps
OUTPUT = ROOT / 'run_analysis' / 'outputs'
evidence_class_findings = evidence_class_effects(ROOT / 'results' / 'GPT-5-mini')
display(evidence_class_findings.head())
print(f'Evidence-class analyses: {len(evidence_class_findings)}; BF10 >= 3: {(evidence_class_findings.bf10 >= 3).sum()}')
```

```python
MAIN_BEHAVIORS = ['deceive', 'distrust', 'strategic aggression', 'non-compliance']
plot_evidence_heatmaps(evidence_class_findings, MAIN_BEHAVIORS, OUTPUT / 'fig5_evidence_classes_main')
```

```python
APPENDIX_BEHAVIORS = sorted(set(evidence_class_findings.behavior_name) - set(MAIN_BEHAVIORS))
plot_evidence_heatmaps(evidence_class_findings, APPENDIX_BEHAVIORS, OUTPUT / 'fig5_evidence_classes_appendix')
```
