# Effect structure by manipulated configuration component — Figure 4 (left)
Shows effect size $\Delta$ by the configuration component manipulated by each hypothesized causal variable $X$.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'run_analysis': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from aerobat.analysis.paper import behavioral_finding_numbers, plot_effect_by_configuration_component
OUTPUT = ROOT / 'run_analysis' / 'outputs'
findings, _ = behavioral_finding_numbers(ROOT / 'results' / 'GPT-5-mini')
```

```python
summary = findings.groupby('var_dimension').agg(hypotheses=('axis_slug','size'), findings_with_BF10_ge_3=('bf10', lambda s: int((s >= 3).sum())), max_BF10=('bf10','max')).reindex(['objective','authority','constraints','situational context','resource','actor','risk & return'])
summary['finding_rate'] = summary.findings_with_BF10_ge_3 / summary.hypotheses
display(summary)
```

```python
fig = plot_effect_by_configuration_component(findings, OUTPUT / 'fig4_effect_by_dimension')
```
