# Generalization analysis across subject agents — Figure 4 (right) and table
Compares effect sizes from different LLM subject agents with the corresponding GPT-5-mini findings reported in the effect landscape for the eight sampled hypotheses shared with the replacement discovery results.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'run_analysis': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from aerobat.analysis.paper import inter_subject_generalization_detail_table, inter_subject_generalization_numbers, inter_subject_generalization_table, plot_inter_subject_generalization
from aerobat.analysis.tables import write_inter_subject_generalization_latex
OUTPUT = ROOT / 'run_analysis' / 'outputs'
OUTPUT.mkdir(parents=True, exist_ok=True)
RESULTS = ROOT / 'results'
SUBJECT_AGENT_RESULTS = {'GPT-5-mini (regenerated)': RESULTS/'GPT-5-mini (regenerated)', 'Gemini-3.1-Pro': RESULTS/'Gemini-3.1-Pro', 'Kimi K2.6': RESULTS/'Kimi-K2.6'}
```

```python
generalization = inter_subject_generalization_table(RESULTS / 'GPT-5-mini', SUBJECT_AGENT_RESULTS)
display(generalization)
display(inter_subject_generalization_numbers(generalization, list(SUBJECT_AGENT_RESULTS)))
```

```python
fig = plot_inter_subject_generalization(generalization, list(SUBJECT_AGENT_RESULTS), OUTPUT / 'fig6_generalization_vs_original')
generalization.to_csv(OUTPUT / 'tabular_generalization.csv', index=False)
generalization_detail = inter_subject_generalization_detail_table(RESULTS / 'GPT-5-mini', SUBJECT_AGENT_RESULTS)
write_inter_subject_generalization_latex(generalization_detail, OUTPUT / 'tabular_generalization.tex')
```
