# Manuscript key-number registry
Recomputes the article registry from stored results and writes it to `run_analysis/outputs/key_numbers.json`. Gate counts come from their owning stage; simulation-round counts include only fidelity-passing simulations handed to Stage 4.


```python
from pathlib import Path
import sys
import pandas as pd
ROOT = Path.cwd().resolve()
if ROOT.name == 'run_analysis': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from aerobat.analysis.paper import manuscript_registry
OUTPUT = ROOT / 'run_analysis' / 'outputs'
OUTPUT.mkdir(parents=True, exist_ok=True)
```

```python
registry = manuscript_registry(ROOT, output_path=OUTPUT / 'key_numbers.json')
display(pd.DataFrame([{'key': key, 'value': item['value']} for key, item in registry.items()]))
print(f"Wrote {len(registry)} key numbers to {OUTPUT / 'key_numbers.json'}")
```
